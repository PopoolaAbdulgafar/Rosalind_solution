# Rosalind Problem 3: Reverse Complement

**Objective:** Generate the reverse complement of a DNA sequence.

```python
file_path = r"C:Downloads/rosalind_revc.txt"
with open(file_path, "r") as file:
    dna_sequence = file.read().strip()
reverse_complement = dna_sequence.translate(str.maketrans("ACGT", "TGCA"))[::-1]
print(reverse_complement)
```
