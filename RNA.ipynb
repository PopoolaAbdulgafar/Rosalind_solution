# Rosalind Problem 2: Transcribing DNA into RNA

**Objective:** Convert a DNA sequence into an RNA sequence by replacing T with U.

```python
file_path = r"C:Downloads/rosalind_rna.txt"
with open(file_path, "r") as file:
    dna_sequence = file.read().strip()
rna_sequence = dna_sequence.replace("T", "U")
print(rna_sequence)
```
