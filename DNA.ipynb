# Rosalind Problem 1: Counting DNA Nucleotides

**Objective:** Count the occurance of A, T, C, G in DNA Sequence

```python
file_path = r"C:Downloads/rosalind_dna.txt"
with open(file_path, "r") as file:
    dna_sequence = file.read().strip()
    nucleotide_counts = {base: dna_sequence.count(base) for base in "ACGT"}
    print(*nucleotide_counts.values())
    
    
```
